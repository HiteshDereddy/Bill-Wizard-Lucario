```python
import cv2
import pytesseract
import glob
import os
import time
import csv

# Path to Tesseract executable (Update based on your system's installation)
pytesseract.pytesseract.tesseract_cmd = '/opt/homebrew/bin/tesseract'

# Function to apply OCR to each image and extract text
def extract_text_from_image(image):
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)  # Convert to grayscale
    text = pytesseract.image_to_string(gray_image)  # Extract text using OCR
    return text

# Function to visualize images and extract text from each folder (which represents a bill)
def process_invoices(image_folders):
    extracted_data = []  # To store extracted data for later export
    for folder in image_folders:
        print(f"\nProcessing Folder: {folder}")
        
        # Load all images in the folder
        image_files = sorted(glob.glob(os.path.join(folder, "*.jpg")) + glob.glob(os.path.join(folder, "*.png")))  # Adjust for different formats

        if not image_files:
            print(f"No images found in folder: {folder}")
            continue
        
        for image_file in image_files:
            # Read the image
            frame = cv2.imread(image_file)
            
            if frame is None:
                print(f"Failed to load image: {image_file}")
                continue
            
            # Apply OCR to extract the text from the image
            try:
                extracted_text = extract_text_from_image(frame)
            except Exception as e:
                print(f"Error processing {image_file}: {e}")
                continue
            
            # Display the extracted text and image
            print(f"Extracted Text from {image_file}: {extracted_text}")
            extracted_data.append({'file': image_file, 'text': extracted_text})  # Store for export
            
            # Display the image for visualization
            cv2.imshow(f"Invoice from {folder}", frame)
            
            # Wait for 500 milliseconds between images, or press 'q' to skip to the next folder
            if cv2.waitKey(500) & 0xFF == ord('q'):
                break

            # Simulate a short pause to mimic video frames
            time.sleep(0.5)  # 500 milliseconds pause

        # Cleanup: close all windows after processing each folder
        cv2.destroyAllWindows()

    # Export extracted data to a CSV file
    export_data(extracted_data)

def export_data(data):
    # Export extracted data to a CSV file
    with open('extracted_data.csv', mode='w', newline='') as file:
        writer = csv.DictWriter(file, fieldnames=['file', 'text'])
        writer.writeheader()
        for entry in data:
            writer.writerow(entry)
    print("Extracted data exported to 'extracted_data.csv'.")

# List of folders representing different invoices (e.g., '1', '2', '3', ... '253')
image_folders = [f'/workspace/Untitled Folder/voce{i}' for i in range(1, 254)]  # Adjust based on your folder names

# Run the processing function on the list of folders
process_invoices(image_folders)
```

```python
import cv2
import os

def extract_frames_from_video(video_path, output_folder):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    cap = cv2.VideoCapture(video_path)
    frame_count = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        # Save frame as JPEG file
        frame_filename = os.path.join(output_folder, f'frame_{frame_count}.jpg')
        cv2.imwrite(frame_filename, frame)
        frame_count += 1

    cap.release()
    print(f"Extracted {frame_count} frames from {video_path} to {output_folder}.")

# Example usage
main_folder = '/workspace/Untitled Folder/voce'
for folder_num in range(1, 300):  # Adjust range as necessary
    video_path = os.path.join(main_folder, f'invoice_video_{folder_num}.mp4')
    output_folder = os.path.join(main_folder, str(folder_num), 'frames')
    
    if os.path.exists(video_path):
        extract_frames_from_video(video_path, output_folder)
    else:
        print(f"Video does not exist: {video_path}. Skipping.")
```

```python
import json
import os
import cv2
import pytesseract
import re

# Specify the path to the Tesseract executable
pytesseract.pytesseract.tesseract_cmd = '/usr/bin/tesseract'

def preprocess_image(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    threshold_img = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)
    return threshold_img

def extract_text_from_frames(frame_folder):
    extracted_texts = {}
    
    for frame_file in os.listdir(frame_folder):
        if frame_file.endswith(".jpg"):
            frame_path = os.path.join(frame_folder, frame_file)
            image = cv2.imread(frame_path)
            
            # Preprocess image
            processed_image = preprocess_image(image)
            
            # OCR
            text = pytesseract.image_to_string(processed_image)
            extracted_texts[frame_file] = text.strip()
    
    return extracted_texts

def extract_relevant_info(text):
    amount_match = re.search(r'Amount\s*[:\-]?\s*\$?\d+\.?\d*', text, re.IGNORECASE)
    invoice_match = re.search(r'Invoice\s*[:\-]?\s*\d+', text, re.IGNORECASE)
    
    return {
        'Amount': amount_match.group() if amount_match else None,
        'Invoice': invoice_match.group() if invoice_match else None
    }

def categorize_texts(texts):
    categorized_data = {
        'amounts': [],
        'invoice_details': [],
        'other': []
    }
    
    for file, text in texts.items():
        info = extract_relevant_info(text)
        
        if info['Amount']:
            categorized_data['amounts'].append({file: info['Amount']})
        elif info['Invoice']:
            categorized_data['invoice_details'].append({file: info['Invoice']})
        else:
            categorized_data['other'].append({file: text})

    return categorized_data

def save_to_json(data, output_file):
    with open(output_file, 'w') as f:
        json.dump(data, f, indent=4)

# Main workflow
main_folder = '/workspace/Untitled Folder/voce'  # Update with your path

for folder_num in range(1, 300):  # Adjust range as necessary
    frame_folder = os.path.join(main_folder, str(folder_num), 'frames')
    
    if os.path.exists(frame_folder):
        texts = extract_text_from_frames(frame_folder)
        categorized_texts = categorize_texts(texts)
        
        # Save categorized texts to a JSON file
        output_file = os.path.join(main_folder, f'categorized_texts_folder_{folder_num}.json')
        save_to_json(categorized_texts, output_file)
        print(f"Saved categorized texts to {output_file}")
    else:
        print(f"No frames found in {frame_folder}. Skipping.")
```

```python
import json
import os
import cv2
import pytesseract

# Specify the path to the Tesseract executable
pytesseract.pytesseract.tesseract_cmd = '/usr/bin/tesseract'

def preprocess_image(image):
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # Apply adaptive thresholding
    thresh = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
                                   cv2.THRESH_BINARY, 11, 2)
    
    # Optionally denoise the image
    denoised = cv2.fastNlMeansDenoising(thresh, None, 30, 7, 21)
    
    return denoised

def extract_text_from_frames(frame_folder):
    extracted_texts = {}
    
    for frame_file in os.listdir(frame_folder):
        if frame_file.endswith(".jpg"):
            frame_path = os.path.join(frame_folder, frame_file)
            image = cv2.imread(frame_path)
            preprocessed_image = preprocess_image(image)
            text = pytesseract.image_to_string(preprocessed_image, config='--psm 6')  # PSM 6 is for block of text
            extracted_texts[frame_file] = text.strip()
    
    return extracted_texts

def categorize_texts(texts):
    categorized_data = {
        'amounts': [],
        'invoice_details': [],
        'other': []
    }
    
    for file, text in texts.items():
        # Use regex to find patterns of amounts or invoice numbers
        if 'Amount' in text or re.search(r'\b\d{1,3}(,\d{3})*(\.\d{2})?\b', text):  # Finding amounts
            categorized_data['amounts'].append({file: text})
        elif 'Invoice' in text or re.search(r'Invoice\s*#?\s*\d+', text):  # Finding invoice number
            categorized_data['invoice_details'].append({file: text})
        else:
            categorized_data['other'].append({file: text})

    return categorized_data

def save_to_json(data, output_file):
    with open(output_file, 'w') as f:
        json.dump(data, f, indent=4)

# Main workflow
main_folder = '/workspace/Untitled Folder/voce'  # Update with your path

for folder_num in range(1, 300):  # Adjust range as necessary
    frame_folder = os.path.join(main_folder, str(folder_num), 'frames')
    
    if os.path.exists(frame_folder):
        texts = extract_text_from_frames(frame_folder)
        categorized_texts = categorize_texts(texts)
        
        # Save categorized texts to a JSON file
        output_file = os.path.join(main_folder, f'categorized_texts_folder_{folder_num}.json')
        save_to_json(categorized_texts, output_file)
        print(f"Saved categorized texts to {output_file}")
    else:
        print(f"No frames found in {frame_folder}. Skipping.")
```

```python
import json
import os
import re
import torch
from transformers import pipeline

def process_with_hugging_face_model(categorized_data):
    # Load a pre-trained model for text classification
    model_name = "dbmdz/distilbert-base-uncased-finetuned-sst-2-english"
    classifier = pipeline("sentiment-analysis", model=model_name)

    # Prepare inputs for the model
    texts = categorized_data['amounts'] + categorized_data['invoice_details'] + categorized_data['other']
    input_texts = [list(item.values())[0] for item in texts]  # Extract the text values

    # Process the texts using the classifier
    results = classifier(input_texts)

    # Print the results
    for text, result in zip(input_texts, results):
        print(f"Text: {text} | Label: {result['label']} | Score: {result['score']:.4f}")

# Example of how to call this function
categorized_texts = {
    'amounts': [{'file1.jpg': 'Amount: $100.00'}],
    'invoice_details': [{'file2.jpg': 'Invoice #12345'}],
    'other': [{'file3.jpg': 'Some other text'}]
}

process_with_hugging_face_model(categorized_texts)
```
